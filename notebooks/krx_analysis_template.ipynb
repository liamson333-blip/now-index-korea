# KRX Data Analysis Template

This notebook demonstrates how to fetch Korean stock data, load it with `pandas`, and inspect the results for index analysis. Use it as a starting point for building a NOW-style Korean market index.

## Setup

Install the repository requirements and import the package modules. Run this notebook from the repository root so relative paths work correctly.

```python
# Install dependencies if needed
# !pip install -r ../requirements.txt

from pathlib import Path
import pandas as pd

from now_index_korea.data import fetch_krx_stock_data, fetch_sample_data
from now_index_korea.index import compute_index
```

## Fetch sample data

This section uses the built-in sample data to demonstrate the package import and index computation. Replace `fetch_sample_data` with `fetch_krx_stock_data` once you are ready to work with real KRX tickers.

```python
sample_path = fetch_sample_data(
    output_path=Path('data/sample_market_data.csv')
)
print('Sample data saved to:', sample_path)
df_sample = pd.read_csv(sample_path)
df_sample
```

```python
index_value = compute_index(sample_path)
print('Sample NOW index value:', index_value)
```

## Fetch KRX stock data

Use `fetch_krx_stock_data` to download historical data from Yahoo Finance for KRX tickers. The output CSV includes OHLC and volume fields.

```python
krx_path = fetch_krx_stock_data(
    ['005930', '000660'],
    period='3mo',
    interval='1d',
    output_path=Path('data/krx_stock_data.csv'),
)
print('KRX data saved to:', krx_path)
df_krx = pd.read_csv(krx_path, parse_dates=['Date'])
df_krx.head()
```
